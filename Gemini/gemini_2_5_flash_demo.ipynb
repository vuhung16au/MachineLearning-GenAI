# Setup Environment

```python
from google import genai
import os
from pathlib import Path
# Try to import userdata for Colab detection
try:
    from google.colab import userdata
    COLAB_AVAILABLE = True
except ImportError:
    COLAB_AVAILABLE = False
def load_env_file(env_path):
    """Load environment variables from .env file"""
    env_vars = {}
    if os.path.exists(env_path):
        with open(env_path, 'r') as f:
            for line in f:
                line = line.strip()
                if line and not line.startswith('#') and '=' in line:
                    key, value = line.split('=', 1)
                    env_vars[key.strip()] = value.strip().strip('"').strip("'")
    return env_vars
GOOGLE_API_KEY = None
if COLAB_AVAILABLE:
    try:
        # Check if running on Colab
        GOOGLE_API_KEY = userdata.get('GOOGLE_API_KEY')
        print("Running on Google Colab, using GOOGLE_API_KEY secret.")
    except:
        print("Running on Google Colab, but GOOGLE_API_KEY secret not found.")
else:
    # Running locally - try to load from .env.local file
    env_file_path = Path.home() / "Documents" / "LangGraph-Credentials" / ".env.local"
    if env_file_path.exists():
        print(f"Running locally, reading API key from {env_file_path}.")
        env_vars = load_env_file(env_file_path)
        GOOGLE_API_KEY = env_vars.get('GOOGLE_API_KEY')
        if not GOOGLE_API_KEY:
            print("GOOGLE_API_KEY not found in .env.local file.")
    else:
        print(f"Running locally, but .env.local file not found at {env_file_path}.")
        print("Please create the file with: GOOGLE_API_KEY=your_api_key_here")
if GOOGLE_API_KEY:
    print("Gemini API key configured successfully.")
else:
    print("Gemini API key not found. Please provide the API key.")
```

# Audio File Processing Demo

## Load Audio File

```python
from google import genai
import time

# Define a prompt to describe the audio content
prompt = "Describe the content of this audio file. Include details about any music, speech, sounds, or other audio elements you can identify."

# !wget https://github.com/vuhung16au/MachineLearning-GenAI/raw/refs/heads/main/Gemini/enigma.mp3

audio_path = "enigma.mp3"

# Select a Gemini model that supports multimodal input (including audio)
client = genai.Client(api_key=GOOGLE_API_KEY)

try:
    # Upload the audio file to Gemini API
    audio_file = client.files.upload(file=audio_path)
    print(f"Uploaded file '{audio_file.display_name}' as: {audio_file.uri}")
    
    # Wait for the file to be processed
    # File processing handled automatically by google.genai
    print(f'Audio file processing complete. State: {audio_file.state}')
    
    # Check if the file processing was successful before generating content
    # Process the uploaded file directly with google.genai
    if audio_file.state == 'ACTIVE':    
        # Generate content based on the prompt and the uploaded audio file
        response = client.models.generate_content(
            model=model,
            contents=[prompt, audio_file]
        )
        
        # Print the generated content
        print("\nGenerated content:")
        print(response.text)
        
        # Clean up the uploaded file
        client.files.delete(audio_file.name)
        print(f"Deleted uploaded file: {audio_file.display_name}")
    else:
        print(f"Audio file processing failed with state: {audio_file.state}")
        
except Exception as e:
    print(f"Error processing audio: {e}")
```

## Audio File Processing Demo


```python
from google.genai import types
import google.genai as genai

# Configure the client with API key
client = genai.Client(api_key=GOOGLE_API_KEY)

try:
    # Read the audio file as bytes
    with open(audio_path, 'rb') as f:
        audio_bytes = f.read()
    
    print(f"Successfully loaded audio file: {audio_path}")
    print(f"Audio file size: {len(audio_bytes)} bytes")
    
    # Generate content using the google.genai library
    response = client.models.generate_content(
        model=model,
        contents=[
            'Describe this audio clip. Include details about any music, speech, sounds, or other audio elements you can identify.',
            types.Part.from_bytes(
                data=audio_bytes,
                mime_type='audio/mp3',
            )
        ]
    )
    
    print("\nGenerated content:")
    print(response.text)
    
except FileNotFoundError:
    print(f"Error: Audio file not found at {audio_path}")
    print("Make sure the audio file was downloaded successfully in the previous steps.")
except Exception as e:
    print(f"Error processing audio with google.genai: {e}")
```

# Document (PDF) Processing Demo

## Load PDF File

```python
import tensorflow as tf
# Download a sample PDF file
pdf_url = "https://arxiv.org/pdf/1706.03762.pdf"
pdf_path = tf.keras.utils.get_file('attention_is_all_you_need.pdf', origin=pdf_url, cache_dir='.')
print(f"PDF downloaded successfully to {pdf_path}")
```

## Process PDF File


```python
import pypdf
# Load the downloaded PDF file
try:
    reader = pypdf.PdfReader(pdf_path)
    # Extract text from the first few pages as an example
    pdf_text = ""
    for page_num in range(min(3, len(reader.pages))): # Process first 3 pages
        pdf_text += reader.pages[page_num].extract_text()
    print("PDF loaded and text extracted from the first few pages.")
except Exception as e:
    print(f"Error loading or processing PDF: {e}")
    pdf_text = None
if pdf_text:
    # Define a prompt
    prompt = "Summarize the key contributions of this research paper based on the provided text."
    # Select a Gemini model that supports multimodal input (text and PDF)
    # The configured model supports text and potentially understands the extracted text content.
    client = genai.Client(api_key=GOOGLE_API_KEY)
    # Use the selected model to generate content based on the prompt and the loaded PDF text
    # The API takes content as a list of parts.
    contents = [prompt, pdf_text]
    response = client.models.generate_content(
        model=model,
        contents=contents
    )    # Print the generated content
    print("\nGenerated content:")
    print(response.text)
else:
    print("Skipping PDF processing due to load or processing error.")
```

# Video Processing Demo

## Load Video File

```python
import tensorflow as tf
# Download a sample MP4 video file
video_url = "https://github.com/vuhung16au/MachineLearning-GenAI/raw/refs/heads/main/Gemini/halong.mp4"
video_path = tf.keras.utils.get_file('halong.mp4', origin=video_url, cache_dir='.')
print(f"Video downloaded successfully to {video_path}")
```

## Describe Video

```python
import subprocess
import os
from PIL import Image
import numpy as np
from google import genai
import time
# Load the downloaded video file path (assuming video_path is already defined from previous steps)
# video_path = "./datasets/halong.mp4" # Uncomment and replace if video_path is not defined
# Create a directory to store extracted frames
frames_dir = "./video_frames"
os.makedirs(frames_dir, exist_ok=True)
# Define the command to extract frames using ffmpeg
# This command extracts one frame every 5 seconds
ffmpeg_command = [
    "ffmpeg",
    "-i", video_path,
    "-vf", "fps=1/5",
    f"{frames_dir}/frame_%04d.png"
]
try:
    # Execute the ffmpeg command
    print(f"Executing ffmpeg command: {' '.join(ffmpeg_command)}")
    subprocess.run(ffmpeg_command, check=True, capture_output=True, text=True)
    print("Frame extraction completed successfully.")
    # Load the extracted frames
    frames = []
    frame_files = sorted([f for f in os.listdir(frames_dir) if f.startswith('frame_') and f.endswith('.png')])
    for frame_file in frame_files:
        img_path = os.path.join(frames_dir, frame_file)
        frames.append(Image.open(img_path))
    print(f"Loaded {len(frames)} frames.")
except FileNotFoundError:
    print("Error: ffmpeg command not found. Please ensure ffmpeg is installed and in your PATH.")
    frames = []
except subprocess.CalledProcessError as e:
    print(f"Error during ffmpeg execution: {e}")
    print(f"stdout: {e.stdout}")
    print(f"stderr: {e.stderr}")
    frames = []
except Exception as e:
    print(f"Error loading extracted frames: {e}")
    frames = []
if frames:
    # Define a prompt related to the video content
    prompt = "Describe the key actions and characters in this video based on the provided frames."
    # Select a Gemini model that supports multimodal input (text and image)
    client = genai.Client(api_key=GOOGLE_API_KEY)
    # Use the selected model to generate content based on the prompt and the extracted frames
    # The API takes content as a list of parts.
    contents = [prompt] + frames
    response = client.models.generate_content(
        model=model,
        contents=contents
    )# Print the generated content
    print("\nGenerated content:")
    print(response.text)
else:
    print("Skipping video processing due to frame extraction or loading error.")
```
